```python
# 1. confirm both files load
import polars as pl

df = pl.read_parquet("../../data/clean/flights_clean.parquet")
throughput = pl.read_parquet("../../data/clean/airport_throughput.parquet")

# only collect when needed
print(df.shape)
print(throughput.shape)

# 2. check top 10 airports have enough data
throughput.group_by("DEST").agg(
    pl.len().alias("slot_count")
).sort("slot_count", descending=True).head(10)

# 3. check non-COVID data only
throughput.filter(pl.col("COVID_FLAG") == 0)["DEST"].value_counts().sort("count", descending=True).head(10)
```

```python
throughput.select(
    pl.col("arrivals").max().alias("max_arrivals"),
    pl.col("arrivals").mean().alias("mean_arrivals"),
    pl.col("arrivals").median().alias("median_arrivals")
)
```
